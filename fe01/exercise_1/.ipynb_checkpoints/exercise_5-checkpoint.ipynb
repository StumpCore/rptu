```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

# Exercise 5: Lognormal

```python
# defining variables
s_0 = 100
b_0 = 100
days_p_year =250
r = 0.01
e_p = 1+r
mu = 0.09
sigma  =0.3
delta = 1/days_p_year
alpha = float(sigma * np.sqrt(delta))
beta = mu * delta
n_stocks = 10
T=250

# generate zero stock frame incl. start value
stock_prices = np.zeros((n_stocks, T+1))
stock_prices[:, 0] = s_0

# generate zero bond frame
bond_prices = np.zeros(T+1)
bond_prices[0] = b_0
```

```python
# stock prices
for i in range(n_stocks):
    for t in range(T):
        rt_plus_1 = np.random.normal(beta, alpha)
        stock_prices[i, t+1] = stock_prices[i, t] * np.exp(rt_plus_1)

# bond prices
for t in range(T):
    rho = np.log(1 + r)
    bond_prices[t+1] = bond_prices[t] * np.exp(rho*delta)

# Plotting
plt.figure(figsize=(12, 6))
for i in range(n_stocks):
    plt.plot(stock_prices[i], label=f'Stock Simulation {i+1}')
plt.plot(bond_prices, label='Bond Price', color='black', linewidth=2)
plt.title('Simulated Stock and Bond Prices over one year')
plt.xlabel('days')
plt.ylabel('p')
plt.legend()
plt.grid(True)
plt.show()
```
